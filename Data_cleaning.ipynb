```python
# Check all unique values in the storage column
df["storage"].unique()
```

```python
# Function to convert storage values into GB
def convert_storage(x):
    # Keep missing values as NaN
    if pd.isna(x):
        return x
    
    # Convert the value to uppercase and remove extra spaces
    x = str(x).upper().strip()
    
    # Convert TB to GB
    if "TB" in x:
        return float(x.replace("TB", "").strip()) * 1024
    
    # Convert GB values to numbers
    if "GB" in x:
        return float(x.replace("GB", "").strip())
    
    # Convert numeric values directly to float
    return float(x)
```

```python
# Apply the convert_storage function to each value in the storage column
df["storage"] = df["storage"].apply(convert_storage)
```

```python
# Import the random module
import random

# List of possible colors for Apple laptops
apple_colors = ["Sky Blue", "Silver", "Starlight", "Midnight"]

# Loop through each row of the DataFrame
for i in df.index:

    # Check if the brand is Apple and the color is missing
    if df.loc[i, "brand"] == "Apple" and pd.isna(df.loc[i, "color"]):

        # Fill the missing color with a random Apple color
        df.loc[i, "color"] = random.choice(apple_colors)
```
