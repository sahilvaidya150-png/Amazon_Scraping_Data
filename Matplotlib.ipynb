```python
import matplotlib.pyplot as plt
import pandas as pd
# Read the laptop data CSV file
df = pd.read_csv(r"C:\Users\Soham\Desktop\Web Scraping Project\Web Scraping Project\amazon_laptops_cleaned.csv")
# Display the DataFrame
df
```

```python
# Count laptops by brand
brand_count = df["brand"].value_counts()

# Create bar chart
plt.figure(figsize=(10, 5))
plt.bar(brand_count.index, brand_count.values)

plt.title("Laptop Count by Brand")
plt.xlabel("Brand")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45)

plt.show()
```

```python
# Price distribution
plt.figure(figsize=(10, 5))
plt.hist(df["price"], bins=20)

plt.title("Laptop Price Distribution")
plt.xlabel("Price")
plt.ylabel("Number of Laptops")

plt.show()
```

```python
# RAM distribution
ram_count = df["ram"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.bar(ram_count.index.astype(str), ram_count.values)

plt.title("RAM Distribution")
plt.xlabel("RAM (GB)")
plt.ylabel("Number of Laptops")

plt.show()
```

```python
# Storage distribution
storage_count = df["storage"].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.bar(storage_count.index.astype(str), storage_count.values)

plt.title("Storage Distribution")
plt.xlabel("Storage (GB)")
plt.ylabel("Number of Laptops")

plt.show()
```

```python


# Convert rating to numeric
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Remove missing values
rating = df["rating"].dropna()

# Plot rating
plt.figure(figsize=(10, 5))
plt.hist(rating, bins=10)

plt.title("Laptop Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Laptops")

plt.show()
```

```python
# Top 10 laptop brands
brand_count = df["brand"].value_counts().head(10)

plt.figure(figsize=(10, 5))
plt.bar(brand_count.index, brand_count.values)

plt.title("Top 10 Laptop Brands")
plt.xlabel("Brand")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45)

plt.show()
```

```python
# Price vs Rating
plt.figure(figsize=(10, 5))
plt.scatter(df["price"], df["rating"])

plt.title("Price vs Rating")
plt.xlabel("Price")
plt.ylabel("Rating")

plt.show()
```

```python
# Average price for each brand
avg_price = df.groupby("brand")["price"].mean().sort_values(ascending=False)

plt.figure(figsize=(10, 5))
plt.bar(avg_price.index, avg_price.values)

plt.title("Average Laptop Price by Brand")
plt.xlabel("Brand")
plt.ylabel("Average Price")
plt.xticks(rotation=45)

plt.show()
```

```python
# Processor distribution
processor_count = df["processor"].value_counts().head(10)

plt.figure(figsize=(10, 5))
plt.bar(processor_count.index, processor_count.values)

plt.title("Top 10 Processors")
plt.xlabel("Processor")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45)

plt.show()
```

```python
# Laptop color distribution
color_count = df["color"].value_counts()

plt.figure(figsize=(10, 5))
plt.bar(color_count.index, color_count.values)

plt.title("Laptop Color Distribution")
plt.xlabel("Color")
plt.ylabel("Number of Laptops")
plt.xticks(rotation=45)

plt.show()
```
